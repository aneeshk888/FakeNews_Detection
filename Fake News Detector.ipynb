```python
data_fake["class"]=0
data_true["class"]=1
```

```python
# Extract last 10 rows and remove last 10 rows dynamically
data_fake_manual_testing = data_fake.tail(10)
data_fake.drop(data_fake.index[-10:], axis=0, inplace=True)

data_true_manual_testing = data_true.tail(10)
data_true.drop(data_true.index[-10:], axis=0, inplace=True)
```

```python
# Step 6: Assigning Classes
data_fake_manual_testing = data_fake.tail(10).copy()
data_fake_manual_testing['class'] = 0

data_true_manual_testing = data_true.tail(10).copy()
data_true_manual_testing['class'] = 1

# Step 7: Merging the Correct DataFrames
data_merge = pd.concat([data_fake_manual_testing, data_true_manual_testing], axis=0)
print(data_merge.head(10))
```

```python
# Step 8: Dropping Unwanted Columns
data = data_merge.drop(['title', 'subject', 'date'], axis=1)
```

```python
import re
import string

def wordopt(text):
    text = text.lower()
    text = re.sub(r'\[.*?\]', '', text)
    text = re.sub(r"\W", " ", text)
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'<.*?>+', '', text)
    text = re.sub(r'[%s]' % re.escape(string.punctuation), '', text)
    text = re.sub(r'\n', '', text)
    text = re.sub(r'\w*\d\w*', '', text)
    return text
```

```python
# Step 10: Applying Function to Text Column and Assigning X and Y
data['text'] = data['text'].apply(wordopt)

x = data['text']
y = data['class']
```

```python
# Step 11: Splitting Data into Training and Testing Sets
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25)

# Step 12: Transforming Text Data Using TF-IDF Vectorization
from sklearn.feature_extraction.text import TfidfVectorizer

vectorization = TfidfVectorizer()
xv_train = vectorization.fit_transform(x_train)
xv_test = vectorization.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression

LR = LogisticRegression()
LR.fit(xv_train, y_train)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split

# Splitting data with stratification to ensure balanced classes
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.25, stratify=y)

# Training the Logistic Regression model
LR = LogisticRegression()
LR.fit(xv_train, y_train)

# Making predictions
pred_lr = LR.predict(xv_test)

# Evaluating the model while handling undefined precision warnings
print("Model Accuracy:", LR.score(xv_test, y_test))
print(classification_report(y_test, pred_lr, zero_division=1))
```

```python
from sklearn.tree import DecisionTreeClassifier

DT = DecisionTreeClassifier()
DT.fit(xv_train, y_train)
```

```python
# Step 15: Making Predictions and Evaluating the DecisionTreeClassifier Model
pred_dt = DT.predict(xv_test)
DT.score(xv_test, y_test)
print(classification_report(y_test, pred_dt))
```

```python
def output_label(n):
    if n == 0:
        return "Fake News"
    elif n == 1:
        return "Not A Fake News"

def manual_testing(news):
    testing_news = {"text": [news]}
    new_def_test = pd.DataFrame(testing_news)
    new_def_test["text"] = new_def_test["text"].apply(wordopt)
    new_x_test = new_def_test["text"]
    new_xx_test = vectorization.transform(new_x_test)
    pred_LR = LR.predict(new_xx_test)
    pred_DT = DT.predict(new_xx_test)

    return print("\n\nLR Prediction: {} \nDT Prediction: {}".format(output_label(pred_LR[0]), output_label(pred_DT[0])))

news = str(input())
manual_testing(news)
```
